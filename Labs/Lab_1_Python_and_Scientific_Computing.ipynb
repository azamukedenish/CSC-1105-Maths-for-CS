
## 1.4 Comparison and Logical Operators

Comparison operators return `True` or `False`.

- `==` equal to
- `!=` not equal to
- `<`, `<=`, `>`, `>=`

Logical operators combine conditions:

- `and`
- `or`
- `not`


````python

x = 12

print(x > 5)
print(x == 12)
print(x != 8)
print(x > 5 and x < 20)
print(x < 0 or x > 10)
print(not x == 12)
````

Python provides several built-in data structures.

### Lists

A list is ordered and mutable.


````python

marks = [67, 74, 82, 91, 58]

print(marks)
print(marks[0])
print(marks[-1])

marks.append(88)
print(marks)

print("Number of marks:", len(marks))
print("Maximum:", max(marks))
print("Minimum:", min(marks))
print("Sum:", sum(marks))
````


## 1.8 Conditional Statements

Conditional statements allow a program to make decisions.

General form:

```python
if condition:
    statements
elif another_condition:
    statements
else:
    statements
```

Python uses **indentation** to define blocks.


````python

mark = 73

if mark >= 80:
    grade = "A"
elif mark >= 70:
    grade = "B"
elif mark >= 60:
    grade = "C"
elif mark >= 50:
    grade = "D"
else:
    grade = "F"

print("Grade:", grade)
````


### Example: Determine Whether an Integer is Even or Odd


````python

number = 19

if number % 2 == 0:
    print(number, "is even")
else:
    print(number, "is odd")
````


## 1.9 Loops

Loops repeat statements.

### `for` Loop

A `for` loop is useful when iterating over a collection or sequence.


````python

for i in range(1, 6):
    print(i)
````


### Example: Sum of the First \(n\) Positive Integers


````python

n = 10
total = 0

for k in range(1, n + 1):
    total += k

print("Sum =", total)
````


### `while` Loop

A `while` loop continues while its condition remains true.


````python

count = 5

while count > 0:
    print(count)
    count -= 1

print("Done!")
````


### `break` and `continue`

- `break` terminates a loop.
- `continue` skips the remainder of the current iteration.


````python

for number in range(1, 11):
    if number == 8:
        break
    if number % 2 == 0:
        continue
    print(number)
````


## 1.10 Functions

Functions make code reusable and easier to test.

A function is defined using `def`.


````python

def greet(name):
    return f"Hello, {name}!"

print(greet("Sarah"))
````


### Function with Multiple Parameters


````python

def rectangle_area(length, width):
    return length * width

print(rectangle_area(8, 5))
````


### Default Arguments


````python

def simple_interest(principal, rate=0.10, time=1):
    return principal * rate * time

print(simple_interest(1_000_000))
print(simple_interest(1_000_000, rate=0.12, time=2))
````


### Example: A Function for a Mathematical Sequence

The arithmetic progression


$a_n = a_1 + (n-1)d$


can be represented as a Python function.


````python

def arithmetic_term(a1, d, n):
    return a1 + (n - 1) * d

for n in range(1, 6):
    print(f"a_{n} =", arithmetic_term(3, 4, n))
````


## 1.11 List Comprehensions

A list comprehension provides a concise way to construct a list.


````python

squares = [x**2 for x in range(1, 11)]
print(squares)

even_squares = [x**2 for x in range(1, 11) if x % 2 == 0]
print(even_squares)
````


## 1.12 Exceptions

An exception is an error detected while a program is running.

`try` and `except` allow a program to handle expected errors.


````python

try:
    value = 10 / 0
except ZeroDivisionError:
    print("Division by zero is not allowed.")
````
